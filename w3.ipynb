```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [1, 4, 9, 16]

plt.figure(figsize=(10, 4))  # width, height of the whole figure
plt.suptitle('Quadratic equation')

# First graph - Line
plt.subplot(1, 2, 1)
plt.plot(x, y, marker='*', color='purple', linewidth=2, linestyle='dotted', label='y = x²')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Line Graph')
plt.xlim(0, 5)
plt.ylim(0, 20)
plt.grid()
plt.legend()

# Second graph - Scatter
plt.subplot(1, 2, 2)
plt.scatter(x, y, color='red', s=100, label='y = x²')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Scatter Plot')
plt.xlim(0, 5)
plt.ylim(0, 20)
plt.grid()
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt
import math

x_values = [x for x in range(1, 20)]

# Function 1
y1 = [5*x**3 + 2*x - 1 for x in x_values]
plt.plot(x_values, y1, marker='o', color='purple', label='5x³+2x-1')

# Function 2
y2 = [-2*x**3 + x**2 + 100 for x in x_values]
plt.plot(x_values, y2, marker='*', color='green', label='-2x³+x²+100')

# Function 3
y3 = [2*math.pi*x + 20 for x in x_values]
plt.plot(x_values, y3, marker='.', color='red', label='2πx+20')

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('All Three Functions')
plt.grid()
plt.legend()
plt.show()
```

Plot the trigonometric COSINE function in interval [−3𝜋,3𝜋], i.e., 𝑦 = cos (𝑥), −3𝜋 ≤ 𝑥 ≤3𝜋

```python
import math
import matplotlib.pyplot as plt

center = (2, 2)
r = 4

x_values = []
y_values = []

angle = 0
while angle <= 360:
    x = center[0] + r * math.cos(math.radians(angle))
    y = center[1] + r * math.sin(math.radians(angle))

    x_values.append(x)
    y_values.append(y)

    angle += 1

plt.axis('equal')

plt.plot(x_values, y_values)
plt.show()
```

Plot a circle with center at (a, b) and radius r. The values of a=b=2, and r=4, using a while loop.

```python
import math
import matplotlib.pyplot as plt

# Circle parameters
a = 2
b = 2
r = 4

# Ellipse parameters
h = 2
k = 2
a_e = 6
b_e = 3

x_circle = []
y_circle = []

x_ellipse = []
y_ellipse = []

angle = 0

while angle <= 360:
    rad = math.radians(angle)

    # Circle
    x_c = a + r * math.cos(rad)
    y_c = b + r * math.sin(rad)
    x_circle.append(x_c)
    y_circle.append(y_c)

    # Ellipse
    x_e = h + a_e * math.cos(rad)
    y_e = k + b_e * math.sin(rad)
    x_ellipse.append(x_e)
    y_ellipse.append(y_e)

    angle += 1

plt.plot(x_circle, y_circle, label="Circle")
plt.plot(x_ellipse, y_ellipse, label="Ellipse")

plt.legend()
plt.grid()

ax =plt.gca()
ax.set_aspect('equal',adjustable='box')

plt.show()
```

Plot the above circle and the ellipse on the same axes. This is give an idea about the intersection points of the two conic sections.

```python
import matplotlib.pyplot as plt

a = 3
p = 5
q = 6

x_values = []
y_values = []

x = -5

while x <= 15:
    y = 4 * a * (x - p)**2 + q

    x_values.append(x)
    y_values.append(y)

    x += 0.1

plt.plot(x_values, y_values, label="Parabola")

plt.scatter(p, q)

plt.show()
```
