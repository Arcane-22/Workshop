```python
hrs = float(input("Enter the number of hours: "))
rate = float(input("Enter rate per hour: "))
if hrs<=40:
  total= hrs*rate
else:
  total= 40*rate + (hrs-40) * (rate * 1.5)
print(total)
```

```python
def orderNumbers (n1, n2):
  if n1 > n2:
    return n2, n1
  else:
    return n1, n2
orderNumbers(6,7)
```

```python
def fib (n):
  a=0
  b=1
  for i in range(n):
    print(a)
    f=a+b
    a=b
    b=f
n= int(input("Enter the number of terms: "))
fib(n)
```

```python
def myPow(x, n):
  i=1
  result= x
  while i<n:
    result *= x
    i+=1
  return result
myPow(5,3)

```

```python
def sumDigits(n):
    r = n
    sum = 0
    while r != 0:
        q = r // 10
        digit = r % 10
        sum += digit
        r = q
    return sum

print(sumDigits(234))
```

```python
def squaredSum(n):
  sum = 0
  for i in n:
    sum += i * i
  print(sum)
squaredSum([2,3,4])

```

```python
def sumDigits(num):
  sum = 0
  for i in range(num+1):
    sum+=i
  return sum
sumDigits(6)
```
